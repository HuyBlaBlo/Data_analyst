## Phần 4: Nhóm dữ liệu (Groupby) và tổng hợp


**Câu 13.** Tính tổng `SALES` và tổng `Profit` theo từng `COUNTRY`. Sắp xếp giảm dần theo tổng `SALES` và cho biết Top 5 quốc gia có doanh số cao nhất.

```python
print(df.groupby(by='COUNTRY')[['SALES','Profit']].sum())
```

**Câu 14.** Tính `SALES` trung bình và `Profit` trung bình theo từng nhóm `DEALSIZE`. Nhận xét: nhóm `Large` có thực sự mang lại lợi nhuận trung bình cao nhất không?

```python
print(df.groupby('DEALSIZE')[['SALES','Profit']].sum())
```

```python
orders_by_status = df.groupby('STATUS')['ORDERNUMBER'].sum()

print('so luong don hang duy nhat theo STATUS : ')
display(orders_by_status)

#  total orders
total_unique_orders = df['ORDERNUMBER'].sum()
print(total_unique_orders)

# total orders cancelled
total_order_cancelled = (df['STATUS'] == 'Cancelled').sum()

total_order_disputed = (df['STATUS'] == 'Disputed').sum()

pct_cancelled = (total_order_cancelled / total_unique_orders) * 100
print(pct_cancelled)

```
