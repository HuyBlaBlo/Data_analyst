## Group 1 — Central Tendency

````python
for col in ['age','fare','sibsp','parch','pclass']:
    mean_, median_, mode_ = df[col].mean(), df[col].median(), df[col].mode()[0]
    print(f'{col}: mean={mean_:.2f}, median={median_:.2f},mode={mode_:.2f}')

print()

print(f'mode sex: {df['sex'].mode()[0]}')
print(f'mode embarked: {df['embarked'].mode()[0]}')
print(f'mode class: {df['class'].mode()[0]}')
print(f'mode who: {df['who'].mode()[0]}')
print(f'mode embark_town: {df['embark_town'].mode()[0]}')
````

## Group 2 — Dispersion

````python
# TODO
for col in ['age','fare']:
    c = df[col]
    # range
    range_ = c.max() - c.min()

    #variance
    var_ = c.var()

    #std
    std_ = c.std()

    #IQR
    q1 = c.quantile(0.25)
    q3 = c.quantile(0.75)
    iqr = q3-q1

    #confficient of variation
    cv_ = std_/c.mean()
    print(f'-----{col}-----')
    print(f'Range = {range_:.2f}, variance = {var_:.2f}, standard deviation = {std_:.2f}, IQR = {iqr}, cv = {cv_}')
````

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

````python
survival_class = (df.groupby('class')['survived'].mean() * 100).sort_values(ascending=False)
display(survival_class.round(2))

best_class = survival_class.index[0]
best_rate = survival_class.iloc[0]

difference = survival_class.loc['First'] - survival_class.loc['Third']

print(f"=> Hạng có tỷ lệ sống sót cao nhất là: {best_class} ({best_rate:.2f}%)")
print(f"=> Chênh lệch giữa first và third là: {difference:.2f}%")
````

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

````python
# sex
sex_survival = df.groupby('sex')['survived'].mean() * 100
difference_sex = sex_survival['female'] - sex_survival['male']

#class
pclass_survival = df.groupby('pclass')['survived'].mean() * 100
difference_pclass = pclass_survival[1] - pclass_survival[3]

#use pviot table
combine_analysis = df.pivot_table(index='sex', columns='pclass', values='survived', aggfunc='mean') * 100

print("---SEX--")
display(sex_survival.round(2))
print(f"=> Chênh lệch giữa nam và nữa: {difference_sex:.2f}%\n")

print("---CLASS---")
display(pclass_survival.round(2))
print(f"=> Chênh lệch giữa hạng 1 và hạng 3: {difference_pclass:.2f}%\n")

print("SEX and CLASS")
display(combine_analysis.round(2))
````

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

````python
fare_stats = df.groupby('survived')['fare'].agg(
    so_luong='count',
    mean='mean',
    nedian='median',
    std='std',
    min='min',
    max='max'
)

display(fare_stats.round(2))

df['fare_group'] = pd.qcut(df['fare'], q=4, labels=['Thấp', 'Trung bình thấp', 'Trung bình cao', 'Cao'])
fare_group_survival = df.groupby('fare_group', observed=False)['survived'].mean() * 100

print("---tỉ lệ soogns sót theo phân khúc giá")
display(fare_group_survival.round(2))
````

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

````python
so_luong = df.groupby('family_size')['survived'].count()
ty_le = (df.groupby('family_size')['survived'].mean() * 100)

family_stats = pd.DataFrame({
    'so_luong': so_luong,
    'ty_le_song_sot (%)': ty_le
})

display(family_stats.round(2))
````

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

````python

so_luong_cang = df.groupby('embark_town')['survived'].count()
ty_le_cang = df.groupby('embark_town')['survived'].mean() * 100

port_survival = pd.DataFrame({
    'so_luong': so_luong_cang,
    'ty_le_song_sot (%)': ty_le_cang
}).sort_values(by='ty_le_song_sot (%)', ascending=False)

display(port_survival.round(2))

best_port = port_survival.index[0]
best_port_rate = port_survival.iloc[0]['ty_le_song_sot (%)']

print(f"=> Cảng có tỷ lệ sống sót cao nhất là: {best_port} ({best_port_rate:.2f}%)")
````
